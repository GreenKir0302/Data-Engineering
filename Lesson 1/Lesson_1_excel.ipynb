# Исследуем энтропию данных


```python
# Для расчета энетропии нам нужны библиотеки, способные работать с формулами
import math
# Определим сколько у нас нарушений
total = len(df)
print('Всего нарушений:', total)
```

```python
# Формула Шеннона
entropy_total = math.log2(total)
print('Общая энтропия', entropy_total)
# Задание № 1 - определить почему формула math.log2(total) не похожа на формулу Шеннона выше.
```

Ответ: это та же формула Шеннона, просто для случая, когда все исходы равновероятны (p = 1/N). Тогда сумма упрощается и получается log2(N). То есть тут каждое нарушение считается отдельным уникальным событием, и это максимальная энтропия.

```python
total = len(df)
entropy_total = math.log2(total)

# энтропия по виду нарушения
counts_type = df["Классификация (Вид)"].value_counts()['count']
entropy_type = -sum((c/total) * math.log2(c/total) for c in counts_type if c > 0)

# энтропия по классу нарушения (класс = группа)
counts_class = df["Классификация (Группа)"].value_counts()['count']
entropy_class = -sum((c/total) * math.log2(c/total) for c in counts_class if c > 0)
```

```python
print(f"\n Энтропия общая: H = {entropy_total:.3f} бит")
print(f"🎯 Энтропия по виду нарушения: H = {entropy_type:.3f} бит")
print(f"📉 Неопределенность снизилась на {entropy_total - entropy_type:.3f} бит "
      f"({(1 - entropy_type/entropy_total)*100:.1f}%)")
print(f"🎯 Энтропия по классу нарушения: H = {entropy_class:.3f} бит")
print(f"📉 Неопределенность снизилась на {entropy_total - entropy_class:.3f} бит "
      f"({(1 - entropy_class/entropy_total)*100:.1f}%)")
```

Вывод: общая энтропия 13.56 бит, по виду 3.81, по классу 3.56. Знание вида убирает примерно 9.76 бит неопределенности (около 72%), класса примерно 10 бит (74%). Информация это и есть эта разница. По классу энтропия меньше, потому что классов меньше, чем видов (104 против 182). Еще видно, что виды встречаются неравномерно, несколько видов попадаются сильно чаще остальных.
